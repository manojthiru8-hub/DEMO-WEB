| Parameter | Meaning               |
| --------- | --------------------- |
| `left`    | Left dataframe        |
| `right`   | Right dataframe       |
| `on`      | Common column to join |
| `how`     | Join strategy         |


| Join    | Meaning                   |
| ------- | ------------------------- |
| `inner` | Only matching records     |
| `left`  | All rows from left table  |
| `right` | All rows from right table |
| `outer` | All rows from both tables |


```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id":[101,102,103],
    "name":["Ravi","Priya","Arun"],
    "city":["Chennai","Bangalore","Hyderabad"]
})

orders = pd.DataFrame({
    "order_id":[1,2,3],
    "customer_id":[101,103,101],
    "amount":[500,700,300]
})

customers
```

```python
orders
```

![](https://media.geeksforgeeks.org/wp-content/uploads/joinimages.png)

```python
pd.merge(left = customers,right = orders,left_on = "customer_id",right_on = "customer_id",)
```

```python
pd.merge(left = customers,right = orders,on = "customer_id")
```

```python
pd.merge(left = customers,right = orders,on = "customer_id",how = "outer")
```

```python
pd.merge(left = customers,right = orders,on = "customer_id",how = "right")
```

```python
pd.merge(left = customers,right = orders,on = "customer_id",how = "left")
```
