## NumPy (Numerical Python)
- it is a fundamental library for numerical computing in Python.
- It provides support for large, multi-dimensional arrays and matrices, along with a collection of high-level mathematical functions to operate on them efficiently.

## Key Features of NumPy
- N-Dimensional Arrays
- At the heart of NumPy is the ndarray object, which is an efficient multi-dimensional array for numeric data.


## Vectorized Operations
NumPy allows operations on arrays without the need for explicit loops, increasing performance.

a = np.array([1, 2, 3])

b = np.array([4, 5, 6])

print(a + b)  # Output: [5 7 9]

print(a * b)  # Output: [4 10 18]

## Broadcasting

- It supports broadcasting, allowing operations between arrays of different shapes in a memory-efficient way.

## Mathematical Functions
- NumPy provides many mathematical functions including sin, cos, log, sqrt, etc.
print(np.sqrt(np.array([1, 4, 9])))  # Output: [1. 2. 3.]

## Linear Algebra Support
- NumPy includes tools for linear algebra such as matrix multiplication, eigenvalues, determinants, etc.
mat = np.array([[1, 2], [3, 4]])
print(np.linalg.det(mat))  # Output: -2.0
Random Number Generation
NumPy has functions to generate random numbers, arrays, normal distributions, and more.
print(np.random.rand(2, 3))  # Example: 2x3 array of random floats between 0 and 1
Integration with Other Libraries
It forms the foundation for many data science libraries like Pandas, SciPy, Matplotlib, and machine learning frameworks like TensorFlow and PyTorch.
Why Use NumPy?
Performance: Operations on NumPy arrays are faster than standard Python lists due to optimized C code behind the scenes.
Memory Efficiency: Arrays consume less memory than Python lists.
Ease of Use: Simplifies handling of vectors, matrices, and large numerical datasets.

## Data wrangling,
- its also known as data munging, is the process of cleaning, transforming, and preparing raw data into a structured and useful format for analysis or modeling.
## Definition and Purpose
- Data wrangling involves taking messy, unstructured, or inconsistent data and converting it into a form suitable for analysis.
- It is a critical step in data science and analytics because most raw datasets contain errors, missing values, inconsistencies, or irrelevant information that can distort insights or predictions.
- The goal is to maximize data quality and usability so that analytical models and reports are reliable.
